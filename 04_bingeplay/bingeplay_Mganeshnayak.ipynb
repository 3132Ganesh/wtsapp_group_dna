```python
#Q8 - The over-paying Premium/Family users
#ANS 366 USER WHOSE ENTIRE WATCH HISTORY IS BASIC-TIER ONLY
```

```python
Q8="""SELECT S.USER_ID FROM SUBSCRIPTIONS S 
JOIN WATCH_SESSIONS W ON S.USER_ID=W.USER_ID
JOIN SHOWS SH ON W.SHOW_ID=SH.SHOW_ID
WHERE S.PLAN IN ('PREMUIM','FAMILY')
GROUP BY S.USER_ID
HAVING MIN(SH.MIN_PLAN) ='BASIC';"""
pd.read_sql(Q8,con=mydb)
```

```python
#Tier 3 - The Killer Questions 
#Q11-Consecutive-week engagement
#ans longest streak:26 ,users with 4+ streaks:1675
```

```python
Q11="""WITH user_weeks AS (
    SELECT DISTINCT 
           user_id,
           YEARWEEK(session_date, 3) AS week_num
    FROM watch_sessions
    WHERE user_id IS NOT NULL
),
streaks AS (
    SELECT 
        user_id,
        week_num,
        ROW_NUMBER() OVER (PARTITION BY user_id ORDER BY week_num) AS rn
    FROM user_weeks
),
grouped AS (
    SELECT 
        user_id,
        (week_num - rn) AS grp,
        COUNT(*) AS streak_length
    FROM streaks
    GROUP BY user_id, (week_num - rn)
)
SELECT 
    COUNT(DISTINCT user_id) AS users_with_4plus_streak,
    MAX(streak_length) AS longest_streak,
    (SELECT user_id FROM grouped ORDER BY streak_length DESC LIMIT 1) AS top_user
FROM grouped
WHERE streak_length >= 4;
"""
pd.read_sql(Q11,con=mydb)
```

```python
Q12="""WITH user_weeks AS (
    SELECT DISTINCT 
           user_id,
           YEARWEEK(session_date, 3) AS week_num
    FROM watch_sessions
)
SELECT 
    user_id,
    MIN(week_num) AS start_week,
    MAX(week_num) AS end_week,
    COUNT(*) AS total_weeks
FROM user_weeks
GROUP BY user_id
HAVING COUNT(*) >= 4
   AND MAX(week_num) - MIN(week_num) + 1 = COUNT(*);

"""
pd.read_sql(Q12,con=mydb)
```
